```python
import numpy as np
import pandas as pd

arr = np.array([10, 20, 30, 40, 50])
print("Array: ", arr)
print("Mean : ", arr.mean())
print("Reshaped 2D \n:", arr.reshape(5, 1))
print("Sliced (index 1 to 3) : ", arr[1:4])

data = {
    "Name": ["keerthi", "sakshi", "damini", "harshita"],
    "Marks": [95, 85, 90, 99],
    "Branch": ["CSE", "ISE", "CSE", "ISE"],
}

df = pd.DataFrame(data)
print(df)
print("\n First 2 rows :\n", df.iloc[0:2])
print("\n Marks Column :\n", df["Marks"])
print("\n Rows where Marks > 70 \n", df[df["Marks"] > 70])
```
